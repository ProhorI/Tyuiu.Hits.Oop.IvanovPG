Требования к базовому классу CreditCard:

• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок действия (ExpiryDate).

• Методы:

o GetInfo(): метод для получения информации о кредитной карте в виде
строки.

```
// Polyglot Notebook C# - Демонстрация полиморфизма и Generic классов

using System;
using System.Collections.Generic;
using System.Linq;

// Базовый абстрактный класс Component
public abstract class Component
{
    public string Name { get; set; }
    public string Manufacturer { get; set; }
    public decimal Price { get; set; }
    public string Model { get; set; }

    protected Component(string name, string manufacturer, decimal price, string model)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
        Model = model;
    }

    // Виртуальный метод GetInfo() - может быть переопределен в производных классах
    public virtual string GetInfo()
    {
        return $"{Name} {Model} | {Manufacturer} | {Price:N0} руб.";
    }

    // Перегрузка оператора ToString()
    public override string ToString()
    {
        return GetInfo();
    }

    // Перегрузка операторов сравнения
    public static bool operator ==(Component comp1, Component comp2)
    {
        if (ReferenceEquals(comp1, comp2))
            return true;
        
        if (comp1 is null || comp2 is null)
            return false;

        return comp1.Name == comp2.Name && 
               comp1.Manufacturer == comp2.Manufacturer && 
               comp1.Model == comp2.Model && 
               comp1.Price == comp2.Price;
    }

    public static bool operator !=(Component comp1, Component comp2)
    {
        return !(comp1 == comp2);
    }

    public override bool Equals(object obj)
    {
        if (obj is Component other)
            return this == other;
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Model, Price);
    }
}

// Интерфейс IInstallable
public interface IInstallable
{
    void Install();
    string GetInstallationInfo();
}

// Класс CPU
public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }
    public string Socket { get; set; }

    public CPU(string name, string manufacturer, decimal price, string model, 
               int cores, double frequency, string socket) 
               : base(name, manufacturer, price, model)
    {
        Cores = cores;
        Frequency = frequency;
        Socket = socket;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"💻 {base.GetInfo()} | {Cores} ядер | {Frequency} ГГц | Сокет: {Socket}";
    }

    // Реализация интерфейса IInstallable
    public void Install()
    {
        Console.WriteLine($"🔧 Устанавливаю процессор {Name} {Model} в сокет {Socket}");
    }

    public string GetInstallationInfo()
    {
        return $"⚡ Процессор {Name} устанавливается в сокет {Socket}. Не забудьте нанести термопасту!";
    }
}

// Класс RAM
public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public string Type { get; set; }
    public double Frequency { get; set; }

    public RAM(string name, string manufacturer, decimal price, string model, 
               int capacity, string type, double frequency) 
               : base(name, manufacturer, price, model)
    {
        Capacity = capacity;
        Type = type;
        Frequency = frequency;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"💾 {base.GetInfo()} | {Capacity} ГБ | {Type} | {Frequency} МГц";
    }

    // Реализация интерфейса IInstallable
    public void Install()
    {
        Console.WriteLine($"🔧 Устанавливаю оперативную память {Name} {Model} в слоты DIMM");
    }

    public string GetInstallationInfo()
    {
        return $"⚡ Оперативная память {Type} устанавливается в слоты DIMM. Убедитесь в совместимости!";
    }
}

// Класс GPU
public class GPU : Component, IInstallable
{
    public int VRAM { get; set; }
    public string MemoryType { get; set; }
    public string Interface { get; set; }

    public GPU(string name, string manufacturer, decimal price, string model, 
               int vram, string memoryType, string interfaceType) 
               : base(name, manufacturer, price, model)
    {
        VRAM = vram;
        MemoryType = memoryType;
        Interface = interfaceType;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"🎮 {base.GetInfo()} | {VRAM} ГБ {MemoryType} | {Interface}";
    }

    // Реализация интерфейса IInstallable
    public void Install()
    {
        Console.WriteLine($"🔧 Устанавливаю видеокарту {Name} {Model} в слот {Interface}");
    }

    public string GetInstallationInfo()
    {
        return $"⚡ Видеокарта устанавливается в слот {Interface}. Подключите дополнительное питание!";
    }
}

// Класс HDD
public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }
    public string FormFactor { get; set; }
    public string Interface { get; set; }

    public HDD(string name, string manufacturer, decimal price, string model, 
               int capacity, string formFactor, string interfaceType) 
               : base(name, manufacturer, price, model)
    {
        Capacity = capacity;
        FormFactor = formFactor;
        Interface = interfaceType;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"💽 {base.GetInfo()} | {Capacity} ГБ | {FormFactor} | {Interface}";
    }

    // Реализация интерфейса IInstallable
    public void Install()
    {
        Console.WriteLine($"🔧 Устанавливаю накопитель {Name} {Model} в {FormFactor} отсек");
    }

    public string GetInstallationInfo()
    {
        return $"⚡ Накопитель устанавливается в {FormFactor} отсек. Подключите кабели {Interface}!";
    }
}

// Класс Computer с Generic методами
public class Computer
{
    public string Name { get; set; }
    private List<Component> components;

    public Computer(string name)
    {
        Name = name;
        components = new List<Component>();
    }

    // Generic метод для добавления компонента
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        Console.WriteLine($"✅ Добавлен: {component.Name} {component.Model}");
        component.Install();
        Console.WriteLine();
    }

    // Generic метод для удаления компонента
    public bool RemoveComponent<T>(T component) where T : Component
    {
        var result = components.Remove(component);
        if (result)
        {
            Console.WriteLine($"❌ Удален: {component.Name} {component.Model}");
        }
        else
        {
            Console.WriteLine($"⚠️ Компонент {component.Name} не найден");
        }
        return result;
    }

    // Метод для получения общей стоимости
    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }

    // Метод для отображения информации о всех компонентах
    public void DisplayComponents()
    {
        Console.WriteLine($"\n🖥️ ===== КОМПЬЮТЕР: {Name} ===== 🖥️");
        Console.WriteLine($"📊 Количество компонентов: {components.Count}");
        Console.WriteLine($"💰 Общая стоимость: {GetTotalPrice():N0} руб.");
        Console.WriteLine("┌" + new string('─', 70) + "┐");
        
        int counter = 1;
        foreach (var component in components)
        {
            Console.WriteLine($"│ {counter}. {component.GetInfo()}");
            counter++;
        }
        Console.WriteLine("└" + new string('─', 70) + "┘\n");
    }

    // Перегрузка оператора + для удобного добавления компонентов
    public static Computer operator +(Computer computer, Component component)
    {
        if (component is IInstallable installable)
        {
            computer.components.Add(component);
            installable.Install();
        }
        return computer;
    }
}

// Статический класс ComputerFactory с Generic методами
public static class ComputerFactory
{
    // Generic метод для создания компьютера с одним компонентом
    public static Computer CreateComputerWithComponent<T>(string computerName, T component) 
        where T : Component, IInstallable
    {
        var computer = new Computer(computerName);
        computer.AddComponent(component);
        return computer;
    }

    // Метод для создания игрового компьютера
    public static Computer CreateGamingComputer(string name)
    {
        var computer = new Computer(name);
        
        Console.WriteLine("🎯 Создаю игровой компьютер...");
        computer.AddComponent(new CPU("Intel Core i9", "Intel", 45000, "14900K", 24, 5.8, "LGA1700"));
        computer.AddComponent(new RAM("G.Skill Trident Z5", "G.Skill", 15000, "RGB 32GB", 32, "DDR5", 6000));
        computer.AddComponent(new GPU("NVIDIA GeForce RTX 4090", "NVIDIA", 120000, "Gaming OC", 24, "GDDR6X", "PCIe 4.0"));
        computer.AddComponent(new HDD("Samsung 980 Pro", "Samsung", 12000, "2TB NVMe", 2000, "M.2", "NVMe"));
        
        return computer;
    }

    // Метод для создания офисного компьютера
    public static Computer CreateOfficeComputer(string name)
    {
        var computer = new Computer(name);
        
        Console.WriteLine("💼 Создаю офисный компьютер...");
        computer.AddComponent(new CPU("Intel Core i5", "Intel", 18000, "13400", 10, 4.6, "LGA1700"));
        computer.AddComponent(new RAM("Corsair Vengeance LPX", "Corsair", 5000, "16GB Black", 16, "DDR4", 3200));
        computer.AddComponent(new GPU("Intel UHD Graphics", "Intel", 0, "730", 0, "Встроенная", "Встроенная"));
        computer.AddComponent(new HDD("WD Blue", "Western Digital", 4000, "1TB HDD", 1000, "3.5\"", "SATA"));
        
        return computer;
    }

    // Метод для создания бюджетного компьютера
    public static Computer CreateBudgetComputer(string name)
    {
        var computer = new Computer(name);
        
        Console.WriteLine("💰 Создаю бюджетный компьютер...");
        computer.AddComponent(new CPU("AMD Ryzen 5", "AMD", 12000, "5600X", 6, 4.6, "AM4"));
        computer.AddComponent(new RAM("Kingston Fury Beast", "Kingston", 3500, "8GB", 8, "DDR4", 3200));
        computer.AddComponent(new GPU("AMD Radeon RX 6600", "AMD", 25000, "XT", 8, "GDDR6", "PCIe 4.0"));
        computer.AddComponent(new HDD("Crucial BX500", "Crucial", 3000, "480GB SSD", 480, "2.5\"", "SATA"));
        
        return computer;
    }
}

// Демонстрационная программа
public class DemoProgram
{
    public static void Run()
    {
        Console.OutputEncoding = System.Text.Encoding.UTF8;
        
        Console.WriteLine("🎯 ===== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА И GENERIC КЛАССОВ ===== 🎯\n");

        // Демонстрация полиморфизма
        Console.WriteLine("1. 📋 ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА");
        Console.WriteLine(new string('═', 60));

        List<Component> components = new List<Component>
        {
            new CPU("AMD Ryzen 7", "AMD", 28000, "7800X3D", 8, 5.0, "AM5"),
            new RAM("Corsair Dominator", "Corsair", 12000, "Platinum RGB 32GB", 32, "DDR5", 5600),
            new GPU("AMD Radeon RX", "AMD", 65000, "7900 XTX", 24, "GDDR6", "PCIe 4.0"),
            new HDD("Seagate FireCuda", "Seagate", 8000, "1TB SSD", 1000, "M.2", "NVMe")
        };

        Console.WriteLine("📦 Список компонентов (полиморфизм в действии):\n");
        foreach (var component in components)
        {
            Console.WriteLine($"   {component.GetInfo()}");
        }

        // Демонстрация перегрузки операторов
        Console.WriteLine("\n2. 🔄 ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ ОПЕРАТОРОВ");
        Console.WriteLine(new string('═', 60));

        var cpu1 = new CPU("Intel Core i7", "Intel", 22000, "13700K", 16, 5.4, "LGA1700");
        var cpu2 = new CPU("Intel Core i7", "Intel", 22000, "13700K", 16, 5.4, "LGA1700");
        var cpu3 = new CPU("AMD Ryzen 7", "AMD", 21000, "7700X", 8, 5.4, "AM5");

        Console.WriteLine($"🔍 Сравнение процессоров:");
        Console.WriteLine($"   cpu1 == cpu2: {cpu1 == cpu2} ✅ (одинаковые модели)");
        Console.WriteLine($"   cpu1 == cpu3: {cpu1 == cpu3} ❌ (разные производители)");
        Console.WriteLine($"   cpu1 != cpu3: {cpu1 != cpu3} ✅ (корректное неравенство)");

        // Демонстрация Generic классов и методов
        Console.WriteLine("\n3. 🎯 ДЕМОНСТРАЦИЯ GENERIC МЕТОДОВ");
        Console.WriteLine(new string('═', 60));

        var myComputer = new Computer("Мой Игровой ПК");

        Console.WriteLine("🛠️ Собираю компьютер с помощью Generic методов:\n");
        myComputer.AddComponent(new CPU("Intel Core i5", "Intel", 15000, "12600K", 10, 4.9, "LGA1700"));
        myComputer.AddComponent(new RAM("Corsair Vengeance", "Corsair", 6000, "RGB Pro 16GB", 16, "DDR4", 3600));
        myComputer.AddComponent(new GPU("NVIDIA GeForce RTX", "NVIDIA", 45000, "4070 Ti", 12, "GDDR6X", "PCIe 4.0"));
        myComputer.AddComponent(new HDD("Kingston NV2", "Kingston", 5000, "1TB NVMe", 1000, "M.2", "NVMe"));

        myComputer.DisplayComponents();

        // Демонстрация фабрики компьютеров
        Console.WriteLine("4. 🏭 ДЕМОНСТРАЦИЯ COMPUTER FACTORY");
        Console.WriteLine(new string('═', 60));

        var gamingPC = ComputerFactory.CreateGamingComputer("💎 Премиум Игровой ПК");
        gamingPC.DisplayComponents();

        var officePC = ComputerFactory.CreateOfficeComputer("📊 Офисная Рабочая Станция");
        officePC.DisplayComponents();

        var budgetPC = ComputerFactory.CreateBudgetComputer("📈 Бюджетный ПК для учебы");
        budgetPC.DisplayComponents();

        // Демонстрация удаления компонентов
        Console.WriteLine("5. 🗑️ ДЕМОНСТРАЦИЯ УДАЛЕНИЯ КОМПОНЕНТОВ");
        Console.WriteLine(new string('═', 60));

        var ramToRemove = new RAM("Corsair Vengeance", "Corsair", 6000, "RGB Pro 16GB", 16, "DDR4", 3600);
        myComputer.RemoveComponent(ramToRemove);
        myComputer.DisplayComponents();

        // Демонстрация интерфейса IInstallable
        Console.WriteLine("6. 🔌 ДЕМОНСТРАЦИЯ ИНТЕРФЕЙСА IINSTALLABLE");
        Console.WriteLine(new string('═', 60));

        IInstallable[] installableComponents = {
            new CPU("AMD Ryzen 5", "AMD", 13000, "7600X", 6, 5.3, "AM5"),
            new GPU("NVIDIA GeForce RTX", "NVIDIA", 32000, "4060 Ti", 8, "GDDR6", "PCIe 4.0")
        };

        Console.WriteLine("📋 Инструкции по установке компонентов:\n");
        foreach (var component in installableComponents)
        {
            Console.WriteLine($"   {component.GetInstallationInfo()}");
            component.Install();
            Console.WriteLine();
        }

        // Итоговая статистика
        Console.WriteLine("7. 📊 ИТОГОВАЯ СТАТИСТИКА");
        Console.WriteLine(new string('═', 60));

        Computer[] allComputers = { myComputer, gamingPC, officePC, budgetPC };
        
        Console.WriteLine("💰 Стоимость всех собранных компьютеров:\n");
        foreach (var computer in allComputers)
        {
            Console.WriteLine($"   {computer.Name}: {computer.GetTotalPrice():N0} руб.");
        }

        var mostExpensive = allComputers.OrderByDescending(c => c.GetTotalPrice()).First();
        var leastExpensive = allComputers.OrderBy(c => c.GetTotalPrice()).First();
        
        Console.WriteLine($"\n🏆 Самый дорогой компьютер: {mostExpensive.Name} - {mostExpensive.GetTotalPrice():N0} руб.");
        Console.WriteLine($"💸 Самый бюджетный компьютер: {leastExpensive.Name} - {leastExpensive.GetTotalPrice():N0} руб.");

        var totalCost = allComputers.Sum(c => c.GetTotalPrice());
        Console.WriteLine($"📈 Общая стоимость всех конфигураций: {totalCost:N0} руб.");

        Console.WriteLine($"\n🎉 ===== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА! ===== 🎉");
    }
}

// Запуск демонстрационной программы
DemoProgram.Run();
```
