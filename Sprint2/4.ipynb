```
using System;
using System.Collections.Generic;

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine("---- Информация о сотруднике ----");
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary:F2}");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        Console.WriteLine("---- Менеджер----");
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Отдел: {Department}");
        Console.WriteLine($"Зарплата: {Salary:F2}");
    }
}

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    private List<string> teamMembers;

    public ProjectManager(string name, string position, double salary, string department, string projectName, IEnumerable<string> team)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        teamMembers = new List<string>(team);
    }

    public override void PrintInfo()
    {
        Console.WriteLine("---- Менеджер проекта ----");
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Отдел: {Department}");
        Console.WriteLine($"Проект: {ProjectName}");
        Console.WriteLine($"Зарплата: {Salary:F2}");
    }

    public string GenerateReport()
    {
        return $"Сгенерированный отчёт'{ProjectName}' от {Name} (Отдел: {Department}). Число участников: {teamMembers.Count}.";
    }

    public string[] GetTeamMembers()
    {
        return teamMembers.ToArray();
    }

    public void AddTeamMember(string member)
    {
        teamMembers.Add(member);
    }
}


Employee emp = new Employee("Иванов Прохор", "разработчик, аналитик", 1000000.0);
emp.PrintInfo();
Console.WriteLine();

Manager mgr = new Manager("Никита Коротков", "Менеджер, асистент", 800000.0, "Зарплата");
mgr.PrintInfo();
Console.WriteLine();

var team = new List<string> { "Прохор", "Никита", "Сергей" };
ProjectManager pm = new ProjectManager("Сергей Грабина", "Менеджер проекта", 950000.0, "", "ВШЦТ", team);
pm.PrintInfo();
Console.WriteLine();

string report = pm.GenerateReport();
Console.WriteLine("Сгенерированный отчет:");
Console.WriteLine(report);
Console.WriteLine();

Console.WriteLine("Команда:");
foreach (var member in pm.GetTeamMembers())
    Console.WriteLine($"- {member}");

pm.AddTeamMember("Даниил");
Console.WriteLine();
Console.WriteLine("Члены команды после добавления одного:");
foreach (var member in pm.GetTeamMembers())
    Console.WriteLine($"- {member}");
```
